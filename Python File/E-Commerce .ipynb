```python
df["age_group"] = pd.cut(df["age"],
    bins=[17, 25, 35, 45, 55, 65, 75],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65", "66-75"])

age_group_count = df["age_group"].value_counts().sort_index()
plt.figure(figsize=(10, 6))
plt.plot(
    age_group_count.index,
    age_group_count.values,
    marker="o")
plt.title("Customer Count by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")
plt.grid(True)
plt.show()
```
